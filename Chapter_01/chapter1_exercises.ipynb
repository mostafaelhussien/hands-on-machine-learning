# Chapter 1 - The Machine Learning Landscape

## Exercises Solutions

**1.** Machine Learning is the science of programming computers so that they can learn from data.

**2.** Four types of problems where Machine Learning shines are:
- Problems that require a lot of hand-tuning or long lists of rules.
- Complex problems for which there is no good traditional manual solution.
- Fluctuating environments where the system needs to adapt to new changing data.
- Gaining insights from large amounts of data to discover hidden patterns.

**3.** A labeled training set is a dataset where each instance has the correct answer (label). This allows the system to learn from these examples. For instance, a spam filter is trained on emails labeled as "spam" or "ham" so it can classify new emails correctly.

**4.** The two most common supervised tasks are **Classification** and **Regression**.  
Classification is used to predict categorical outcomes (e.g., spam or not spam), while Regression is used to predict numerical values (e.g., house price).  
Note that some regression algorithms can also be used for classification tasks, such as Logistic Regression.

**5.** Four common unsupervised tasks are:
- Clustering (grouping similar instances)
- Anomaly detection
- Dimensionality reduction
- Visualization of high-dimensional data

**8.** Spam detection is a supervised learning problem. The system is trained on labeled examples of both spam and ham emails. Based on these labeled examples, it learns to classify new emails as spam or not.

**9.** An online learning system is fed with training data sequentially, either one instance at a time or in small mini-batches. This allows the system to learn quickly and with low computational cost. It is very useful for systems that receive continuous data streams and for handling large datasets that do not fit in memory (out-of-core learning). However, it can be sensitive to bad data since the learning happens live.

**10.** Out-of-core learning refers to training a machine learning system on a very large dataset that does not fit entirely into the computer's main memory. The algorithm loads and processes the data incrementally in small batches. Although it is often performed offline, it is closely related to online learning techniques.

**11.** Model parameters are values that are adjusted during training so the model can fit the data well. These parameters are learned by the model itself. On the other hand, hyperparameters are set by the user before training and remain constant during training. They control aspects such as the strength of regularization to avoid overfitting or underfitting.

**12.** 
- **Instance-based learning**: The system learns the training examples by heart and generalizes to new cases by comparing them to similar stored examples using a similarity measure.
- **Model-based learning**: The system builds a mathematical model from the training examples by adjusting its parameters. It uses a cost function (for example in linear regression) to measure the distance between predictions and actual values and tries to minimize it.

**13.** 
- **Overfitting**: Occurs when the model is too complex relative to the amount of training data and noise. It performs very well on the training data but fails to generalize to new data. The solution is usually to increase regularization or reduce the model's degree of freedom.
- **Underfitting**: Occurs when the model is too simple to capture the underlying structure of the data. The solution is to reduce regularization, use a more powerful model, or perform better feature engineering.

**14.** The “No Free Lunch” theorem states that there is no single model that is guaranteed to be the best for all problems. No model is universally superior to others.
